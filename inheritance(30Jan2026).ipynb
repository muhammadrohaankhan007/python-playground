```python
class Animal():
    def __init__(self, typ, legs, color, sound):
        # instance attributes
        self.typ = typ
        self._legs = legs
        self.color = color
        self.sound = sound
    # instance methods
    def get_legs(self):
        return f"I have {self._legs} legs"
    def set_legs(self,new_legs):
        if new_legs>4:
            return "I cant have more than 4 legs"
        self._legs = new_legs
        return f"Legs updated successfully. New legs are {self._legs}"
    
    def eat(self):
        if self.typ =="carnivore":
            return "Animal is eating meat"
        else:
            return "Animal is eating grass"
    def make_sound(self):
        return self.sound
```

```python
class Lion(Animal):
    def __init__(self, typ, legs, color, sound, character, speed, location):
        # self.typ = typ
        # self._legs = legs these are being inherited from above!
        # self.color = color
        # self.sound = sound
        super().__init__(typ,legs, color, sound)      #super ke zaryeh hu =m parents se attributes import kr skte hain jo ke alr overwrite hogayen!
        self.character=character
        self.speed=speed
        self.location=location
    def hunt(self):
        return f"Our {self.color} lion of {self._legs} legs is hunting in {self.location}"
   
        
```

```python
li=Lion("carnivore",4,"brown","roarrrrrrr","mammy daddy","80km/h", "forest")
```

```python
li.hunt()
```

```python
class Car():
    def __init__(self, transmission, color, brand, model, body):
        self.transmission = transmission
        self.color = color
        self.brand = brand
        self.model = model
        self.body = body

    def car_type(self):
        return f"Your car type is {self.body}"

    def car_transmission(self):
        return f"Your car transmission is {self.transmission}"

    def car_color(self):
        return f"Your car color is {self.color}"

    def car_brand(self):
        return f"Your car brand is {self.brand}"

    def car_model(self):
        return f"Your car model is {self.model}"

    def summary(self):
        return (
            f"Your {self.brand} {self.model} is a {self.color} "
            f"{self.body} with {self.transmission} transmission."
        )
```

```python
c1=Car("auto", "brown","ford","raptor", "pickup truck")
```

```python
c1.summary()
```

```python
class Truck(Car):
    def __init__(self, transmission, color, brand, model, body, length, weight):
        super().__init__(transmission, color, brand, model, body)
        self.length=length
        self.weight=weight
    def truck_length(self):
        return f" you have passed the length, hence its a truck and its length i {self.length} with a loading capacity of {self.weight}"
```

```python
t1=Truck("auto","yellow","peterbelt","m23", "truck","100m","2000KG")
```

```python
t1.truck_length()
```
