```python
import geopandas as gpd

# Path to your extracted Geofabrik GeoPackage file
gpkg_path = "new-york-latest-free.gpkg"

# List available layers in the GeoPackage to confirm the railway layer name
layers = gpd.io.file.fiona.listlayers(gpkg_path)
print("Available layers:", layers)

# Typically, lines or railways are stored in a layer like 'lines' or 'railways'
# Let's load the lines layer and filter for railways
lines = gpd.read_file(gpkg_path, layer='lines')

# Filter for features where the railway tag is present
railways = lines[lines['railway'].notnull()]

print(f"Loaded {len(railways)} railway segments in New York.")
print(railways.head())

# Optional: Save to a smaller GeoJSON if needed for your frontend map
# railways.to_file("ny_railways.geojson", driver="GeoJSON")
```
