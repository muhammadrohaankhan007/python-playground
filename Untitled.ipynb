```python
import math

f = float(input("Enter force: "))
d = float(input("Enter distance: "))
theta = float(input("Enter theta (in degrees): "))

theta_rad = math.radians(theta)

W = f * d * math.cos(theta_rad)

print("Work done:", W)
```

```python
m = float(input("Enter mass (kg): "))
c = float(input("Enter specific heat capacity (J/kg°C): "))
T_initial = float(input("Enter initial temperature (°C): "))
T_final = float(input("Enter final temperature (°C): "))

delta_T = T_final - T_initial
Q = m * c * delta_T

print("Heat transferred (Q):", Q, "Joules")
```
