```python
all_accounts=[]
import random
```

```python
def open_account(cnic, title, initial_deposit):
    
    account={
    "cnic": cnic,
    "title":title,
    "balance":initial_deposit,
    "account_number":random.randint(1001,9999),
    "pin":random.randint(1001,9999)
     }

    all_accounts.append(account)

    return f"""Your Account has been Created Successfully!
                ---- Imp info ----
                Account Number: {account['account_number']}
                Pin: {account['pin']}"""


```

```python
open_account("42501-2552030-9", "muhamamd", 1000)
```

```python
all_accounts
```

```python
def check_balance (account_number, pin):
    for acc in all_accounts:
        if acc["account_number"] == account_number:
            if acc["pin"] == pin:
                   return f" balance: {acc["balance"]}"
            else:
                return f" pin {pin_number} not found!"
    else:
        return f" Account {account_number} not found!"
```

```python
check_balance(3722, 3844)
```

```python
print(check_balance(3722, 3844))
```

```python
def with_draw(account, pin, amount):
    for acc in all_accounts:
        if acc["account_number"] == account:
            if acc["pin"] == pin:
                if acc["balance"] >= amount:
                    acc["balance"]-=amount
                    return f""" Withdrawal Successful!
                                {acc["balance"]} withdrawed from account
                                Remining amount: {acc["balance"]}"""
                else:
                     return f" insufficient amount"
            else:
                return f" Account {pin} is incorrect"
    else:
        return f" Account {account} not found"
```

```python
with_draw(3722, 3844, 100)
```

```python
def deposit(account, amt):
    for acc in all_accounts:
        if acc["account_number"]==account:
            acc["balance"]+=amt
            return f""" deposit Successfully, Newer balance: {acc["balance"]} """
    else:
        return f" {account} not found"
```

```python
deposit(3722, 100)
```

```python
def transfer(account, pin ,amt, bacc):
    for ba in all_accounts:
        if ba['account_number']==bacc:        
            for acc in all_accounts:
                if acc['account_number']==account:
                    if acc['pin']==pin:
                        if acc['balance']>=amt:
                            ba['balance']+=amt
                            acc['balance']-=amt
                            return f'{amt} transfered successfully.'                          
                           
                        else:
                            return f"Insufficient Amount"
                    else:
                        return f"Invalid Pin"                
            else:        
                return f"Invalid Account Number" 
    else:
        return f"Invalid Beneficiary Account Number"
```

```python
def change_pin(account, pin, new_pin):
    for acc in all_accounts:
        if acc["account_number"]==account:
            if acc["pin"]==pin:
                acc["pin"]=new_pin
                return f"Pin Changed Sucessfully"
            else:
                return f"pin not found"
    else:
        return f"invalid Account number"
```
