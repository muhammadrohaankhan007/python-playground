```python
print("Decimal into Binary: ")
num = int(input("enter the number: "))
binary=""
if num == 0:
    print("Result: 0")
else: 
    while num > 0:
        reminder=num%2
        binary=str(reminder) + binary
        num=num//2
    print("Result: ", binary)
```

```python
print("Binary Addition using two Variables")
a=input("Enter the first binary number: ")
b=input("Enter the Second Binary Number: ")
sum=bin(int(a,2) + int(b,2))[2:]
print("Result: ", sum)
```

```python
print("Ones Compliment")
binary=input("Enter the Compliment number: ")
ones=""
for bits in binary:
    if bits == "1":
        ones+="0"
    else:
        ones+="1"
print("Result: ", ones)
```

```python
print("Ones Complement")
binary = input("Enter the binary number: ")

if not all(bit in "01" for bit in binary):
    print("Invalid binary number!")
else:
    ones = ""
    for bit in binary:
        if bit == "0":
            ones += "1"
        else:
            ones += "0"

    print("Result:", ones)
```

```python
print("Two's Compliment")
binary=input("Enter the Compliment number: ")
ones=""
for bits in binary:
    if bits=="1":
        ones+="0"
    else:
        ones+="1"
result=bin(int(ones,2) + 1) [2:]
print("Result: ", result)
```

```python
print("Two's Complement")
binary = input("Enter the binary number: ")

if not all(bit in "01" for bit in binary):
    print("Invalid binary number!")
else:
    ones = ""
    for bit in binary:
        if bit == "1":
            ones += "0"
        else:
            ones += "1"
            
    result = bin(int(ones, 2) + 1)[2:]
    print("Result:", result)
```

```python
print("Binary to Hexadecimal")
binary=input("Enter the Binary Number: ")
decimal=int(binary,2)
hex_value=hex(decimal) [2:]
print("Hexadecimal: ", hex_value)
```

```python
print("binary into octal")
binary=input("Enter the Binary Number: ")
decimal=int(binary,2)
octal_value=oct(decimal) [2:]
print("Octal Value: ", octal_value)
```
