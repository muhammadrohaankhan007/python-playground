```python
def clean_email(email):
    if email.endswith("@"):
        return email+"gmail.com"
    elif email.endswith("None"):
        return email.replace("None","gmail.com")
    else:
        return email         
```

```python
df["Email"]=df['Email'].apply(clean_email)
```
